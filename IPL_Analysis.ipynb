```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
matches = pd.read_csv('matches.csv')
deliveries = pd.read_csv('deliveries.csv', on_bad_lines='skip')
```

```python
matches.head()
deliveries.head()
```

```python
top_batsman = deliveries.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10)
top_batsman
```

```python
top_batsman.plot(kind='bar')
plt.title("Top 10 Batsman by Runs")
plt.xlabel("Player")
plt.ylabel("Runs")
plt.xticks(rotation=45)
plt.show()
```

```python
top_bowlers = deliveries[deliveries['is_wicket'] ==1]['bowler'].value_counts().head(10)
top_bowlers
```

```python
top_bowlers.plot(kind='bar')
plt.title("Top 10 Bowlers by wickets")
plt.xlabel("Bowler")
plt.ylabel("Wickets")
plt.xticks(rotation=45)
plt.show()
```

```python
top_teams = matches['winner'].value_counts().head(10)
top_teams
```

```python
top_teams.plot(kind='bar')
plt.title("Top Teams by Wins ")
plt.xlabel("Teams")
plt.ylabel("Wins")
plt.xticks(rotation=45)
plt.show()
```
